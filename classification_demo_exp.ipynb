```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report, accuracy_score
```

```python
# read data after "pdf_corpus_to_csv.R" script completing
docs = pd.read_csv(
    "df2.csv",
    sep=";",
    encoding="windows-1251"
)
```

```python
docs = docs.dropna(subset=["text", "class"]).copy()
docs["text"] = docs["text"].astype(str)
```

```python
# split data before vectorizator learning
train, test = train_test_split(
    docs,
    test_size=0.25,
    random_state=42,
    stratify=docs["class"]
)
```

```python
# vectorization: word and symbolic features can be checked separately;
# start with the word ones.
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.98,
    sublinear_tf=True,
    dtype="float32"
)
```

```python
X_train = vectorizer.fit_transform(train["text"])
X_test = vectorizer.transform(test["text"])

y_train = train["class"]
y_test = test["class"]
```

```python
# classificator learning
model = LinearSVC(
    C=1.0,
    random_state=42
)

model.fit(X_train, y_train)
pred = model.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, pred))
print(classification_report(
    y_test,
    pred,
    zero_division=0
))
```
